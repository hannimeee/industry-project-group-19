# Preview portfolio output files

Display the first five rows of the final portfolio CSV and Parquet outputs.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

root = Path.cwd()
if not (root / "handoff" / "final").is_dir():
    root = root.parent

csv_path = root / "handoff" / "final" / "portfolio_dashboard.csv"
parquet_path = root / "handoff" / "final" / "portfolio_dashboard.parquet"

print("CSV output — first 5 rows")
display(pd.read_csv(csv_path, dtype={"merchant_abn": "string"}, nrows=5))

print("Parquet output — first 5 rows")
display(pd.read_parquet(parquet_path).head(5))

print("Finished.")
```
